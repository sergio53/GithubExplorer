```python
# from get_jupyter_root_dir import get_jupyter_root_dir; get_jupyter_root_dir()
import jupyter_core
import os,glob,json

def get_jupyter_root_dir():
  jupyter_runtime_dir = jupyter_core.paths.jupyter_runtime_dir()
  json_files = glob.glob(f"{jupyter_runtime_dir}/nbserver-*.json")
  json_files = filter(os.path.isfile, json_files)
  list_of_files = sorted(json_files, key = os.path.getmtime)
  
  with open(list_of_files[-1]) as json_file:
    get_jupyter_root_dir.json = json.load(json_file)    
  root_dir = get_jupyter_root_dir.json['notebook_dir']
  return root_dir
```
